# BEST-Rec v4.0: Text-Enhanced DeepFM

**Complete rewrite.** Replaces transformer/cross-attention with Factorization Machine + DNN.

| Problem in v3 | Fix in v4 |
|---|---|
| Transformer too complex for 1.1 interactions/user | DeepFM: handles sparse data natively |
| Standard KFold = cold for 80%+ of users | Per-user leave-one-out for warm eval |
| 5 loss terms confusing optimizer | Single MSE loss |
| 50M+ params, massive overfitting | ~5M params, proper regularization |
| MAE 1.01 (worse than item-mean 0.92) | Target: warm MAE 0.55-0.70 |

## 3. Data Loading & Feature Preparation

````python
def cached_pkl(name, fn, d=V4_CACHE, force=False):
    p = os.path.join(d, name)
    if os.path.exists(p) and not force:
        print(f'  Cache: {name}'); return pickle.load(open(p, 'rb'))
    print(f'  Compute: {name}...')
    r = fn(); os.makedirs(os.path.dirname(p), exist_ok=True)
    pickle.dump(r, open(p, 'wb')); return r

def cached_pt(name, fn, d=V4_CACHE, force=False):
    p = os.path.join(d, name)
    if os.path.exists(p) and not force:
        print(f'  Cache: {name}'); return torch.load(p, weights_only=True)
    print(f'  Compute: {name}...')
    r = fn(); os.makedirs(os.path.dirname(p), exist_ok=True)
    torch.save(r, p); return r

# Load raw data
raw_path = os.path.join(CACHE_DIR, 'raw_data.pkl')
assert os.path.exists(raw_path), f'Run v2 notebook first to create {raw_path}'
data = pickle.load(open(raw_path, 'rb'))
interactions = data['interactions']
item_meta = data['item_metadata']
N_USERS = data['num_users']
N_ITEMS = data['num_items']
print(f'Users: {N_USERS:,} | Items: {N_ITEMS:,} | Interactions: {len(interactions):,}')
print(f'Avg interactions/user: {len(interactions)/N_USERS:.2f}')

# Item title embeddings (reuse v3 sbert cache if available)
v3_sbert = os.path.join(V3_CACHE, 'item_title_sbert.pt')
if os.path.exists(v3_sbert):
    print(f'  Reusing v3 sbert cache')
    item_title_emb = torch.load(v3_sbert, weights_only=True)
else:
    sbert = SentenceTransformer(SBERT_MODEL, device=str(device))
    def _enc_titles():
        titles = [item_meta[i]['title'] for i in range(N_ITEMS)]
        e = sbert.encode(titles, batch_size=512, show_progress_bar=True, convert_to_numpy=True)
        return torch.tensor(e, dtype=torch.float32)
    item_title_emb = cached_pt('item_title_sbert.pt', _enc_titles)

# Item numeric features
def _item_num():
    a = torch.tensor([item_meta[i]['avg_rating'] for i in range(N_ITEMS)])
    r = torch.tensor([item_meta[i]['rating_num'] for i in range(N_ITEMS)], dtype=torch.float32)
    p = torch.tensor([item_meta[i]['price'] for i in range(N_ITEMS)])
    def z(x): return (x - x.mean()) / (x.std() + 1e-9)
    return torch.stack([z(a), z(r), z(p)], dim=-1)
item_num_feat = cached_pt('item_numeric.pt', _item_num)

print(f'Item title embeddings: {item_title_emb.shape}')
print(f'Item numeric features: {item_num_feat.shape}')
````

## 5. Evaluation Splits

**Warm (per-user leave-one-out):** For each user with 2+ interactions, hold out 1 random interaction. The rest go to training. This guarantees user features exist at test time.

**Cold (GroupKFold):** Hold out entire users. Pure cold-start.

````python
def make_warm_split(interactions, seed=SEED):
    """Per-user leave-one-out: hold out 1 interaction per user with 2+ interactions."""
    rng = np.random.RandomState(seed)
    user_inters = defaultdict(list)
    for idx, inter in enumerate(interactions):
        user_inters[inter['user_id']].append(idx)

    train_idx, test_idx = [], []
    for uid, indices in user_inters.items():
        if len(indices) == 1:
            train_idx.extend(indices)  # single-interaction users: train only
        else:
            rng.shuffle(indices)
            test_idx.append(indices[0])      # hold out 1
            train_idx.extend(indices[1:])     # rest to train

    return np.array(train_idx), np.array(test_idx)


def make_warm_kfold(interactions, n_splits=NUM_FOLDS, seed=SEED):
    """K-fold per-user split: for users with k+ interactions, distribute across folds."""
    rng = np.random.RandomState(seed)
    user_inters = defaultdict(list)
    for idx, inter in enumerate(interactions):
        user_inters[inter['user_id']].append(idx)

    # Users with only 1 interaction always go to train
    always_train = []
    foldable = defaultdict(list)  # uid -> list of indices
    for uid, indices in user_inters.items():
        if len(indices) == 1:
            always_train.extend(indices)
        else:
            rng.shuffle(indices)
            foldable[uid] = indices

    # For each foldable user, assign 1 interaction to each fold (round-robin)
    fold_test = [[] for _ in range(n_splits)]
    fold_extra_train = [[] for _ in range(n_splits)]  # remaining interactions

    for uid, indices in foldable.items():
        # Put 1 in test for each fold (cycling), rest in train
        for f in range(n_splits):
            if f < len(indices):
                fold_test[f].append(indices[f])
            # Everything except the test item for this fold goes to train

    splits = []
    for f in range(n_splits):
        te = np.array(fold_test[f])
        te_set = set(fold_test[f])
        # Train = always_train + all foldable indices NOT in this fold's test
        tr = list(always_train)
        for uid, indices in foldable.items():
            for idx in indices:
                if idx not in te_set:
                    tr.append(idx)
        splits.append((np.array(tr), te))

    return splits


# Prepare warm splits
warm_splits = make_warm_kfold(interactions)
for i, (tr, te) in enumerate(warm_splits):
    print(f'  Warm fold {i}: {len(tr):,} train / {len(te):,} test')

# Prepare cold splits
user_ids_arr = np.array([i['user_id'] for i in interactions])
indices_all = np.arange(len(interactions))
cold_gkf = GroupKFold(n_splits=NUM_FOLDS)
cold_splits = list(cold_gkf.split(indices_all, groups=user_ids_arr))
for i, (tr, te) in enumerate(cold_splits):
    print(f'  Cold fold {i}: {len(tr):,} train / {len(te):,} test')
````

## 6. Per-Fold Feature Engineering (Leak-Free)

````python
sbert = None  # lazy init

def get_sbert():
    global sbert
    if sbert is None:
        sbert = SentenceTransformer(SBERT_MODEL, device=str(device))
    return sbert


def encode_user_reviews(train_inters, n_users):
    """Encode user reviews into a single 384-dim vector via attention pooling.
    Only uses TRAINING reviews (leak-free)."""
    model = get_sbert()
    ur = defaultdict(list)
    for i in train_inters:
        if i['review'].strip():
            ur[i['user_id']].append(i['review'])

    # Encode all reviews
    trips = []
    for uid in range(n_users):
        for rev in ur.get(uid, [])[:MAX_REVIEWS]:
            trips.append((uid, rev))

    user_emb = torch.zeros(n_users, SBERT_DIM)
    user_has_text = torch.zeros(n_users, dtype=torch.bool)

    if trips:
        texts = [t[1] for t in trips]
        enc = model.encode(texts, batch_size=512, show_progress_bar=True, convert_to_numpy=True)
        enc = torch.tensor(enc, dtype=torch.float32)

        # Aggregate per user (mean of review embeddings)
        user_sum = torch.zeros(n_users, SBERT_DIM)
        user_count = torch.zeros(n_users)
        for idx, (uid, _) in enumerate(trips):
            user_sum[uid] += enc[idx]
            user_count[uid] += 1

        mask = user_count > 0
        user_emb[mask] = user_sum[mask] / user_count[mask].unsqueeze(1)
        user_has_text = mask

    return user_emb, user_has_text


def compute_fold_features(train_inters, tag, force=False):
    fold_dir = os.path.join(V4_CACHE, tag)
    os.makedirs(fold_dir, exist_ok=True)

    # Item SVD
    def _isvd():
        r, c, v = [], [], []
        for i in train_inters: r.append(i['item_id']); c.append(i['user_id']); v.append(i['rating'])
        mat = csr_matrix((v, (r, c)), shape=(N_ITEMS, N_USERS))
        k = min(SVD_K, min(N_ITEMS, N_USERS)-1, len(set(r))-1)
        if k < 1: return torch.zeros(N_ITEMS, SVD_K)
        svd = TruncatedSVD(n_components=k, random_state=SEED)
        res = svd.fit_transform(mat)
        if k < SVD_K: res = np.hstack([res, np.zeros((N_ITEMS, SVD_K-k))])
        return torch.tensor(res, dtype=torch.float32)
    item_svd = cached_pt('item_svd.pt', _isvd, d=fold_dir, force=force)

    # User SVD
    def _usvd():
        r, c, v = [], [], []
        for i in train_inters: r.append(i['user_id']); c.append(i['item_id']); v.append(i['rating'])
        mat = csr_matrix((v, (r, c)), shape=(N_USERS, N_ITEMS))
        k = min(USER_SVD_K, min(N_USERS, N_ITEMS)-1, len(set(r))-1)
        if k < 1: return torch.zeros(N_USERS, USER_SVD_K)
        svd = TruncatedSVD(n_components=k, random_state=SEED)
        res = svd.fit_transform(mat)
        if k < USER_SVD_K: res = np.hstack([res, np.zeros((N_USERS, USER_SVD_K-k))])
        return torch.tensor(res, dtype=torch.float32)
    user_svd = cached_pt('user_svd.pt', _usvd, d=fold_dir, force=force)

    # User text
    def _ut():
        e, m = encode_user_reviews(train_inters, N_USERS)
        return {'emb': e, 'mask': m}
    ut = cached_pkl('user_text.pkl', _ut, d=fold_dir, force=force)

    # Biases from training data
    gs, gn = 0.0, 0
    us, uc_d = defaultdict(float), defaultdict(int)
    ist, ic_d = defaultdict(float), defaultdict(int)
    for i in train_inters:
        r = i['rating']; gs += r; gn += 1
        us[i['user_id']] += r; uc_d[i['user_id']] += 1
        ist[i['item_id']] += r; ic_d[i['item_id']] += 1
    gm = gs / gn if gn > 0 else 3.0

    ub = torch.zeros(N_USERS)
    ib = torch.zeros(N_ITEMS)
    for uid in range(N_USERS):
        if uc_d[uid] > 0: ub[uid] = (us[uid] / uc_d[uid]) - gm
    for iid in range(N_ITEMS):
        if ic_d[iid] > 0: ib[iid] = (ist[iid] / ic_d[iid]) - gm

    return {
        'item_svd': item_svd, 'user_svd': user_svd,
        'user_text': ut['emb'], 'user_has_text': ut['mask'],
        'gm': gm, 'ub': ub, 'ib': ib,
    }

print('Feature engineering ready.')
````

## 7. Dataset

````python
class RecDS(Dataset):
    def __init__(self, inters, feats):
        self.uids = torch.tensor([i['user_id'] for i in inters], dtype=torch.long)
        self.iids = torch.tensor([i['item_id'] for i in inters], dtype=torch.long)
        self.rats = torch.tensor([i['rating'] for i in inters], dtype=torch.float32)
        self.u_text = feats['user_text']       # (N_USERS, 384)
        self.u_svd = feats['user_svd']         # (N_USERS, 128)
        self.i_title = item_title_emb          # (N_ITEMS, 384)
        self.i_num = item_num_feat             # (N_ITEMS, 3)
        self.i_svd = feats['item_svd']         # (N_ITEMS, 128)

    def __len__(self): return len(self.rats)

    def __getitem__(self, idx):
        uid, iid = self.uids[idx], self.iids[idx]
        return (uid, iid,
                self.u_text[uid], self.u_svd[uid],
                self.i_title[iid], self.i_num[iid], self.i_svd[iid],
                self.rats[idx])

print('Dataset ready.')
````

## 8. BEST-Rec v4: DeepFM Model

```
7 Feature Fields (each 64-dim):
  user_cf, user_text, user_svd, item_cf, item_text, item_svd, item_num

FM: efficient pairwise interactions -> scalar
DNN: concat(all fields) -> MLP(448->256->128->1) -> scalar
Bias: global_mean + user_bias + item_bias

pred = clamp(bias + FM + DNN, 1, 5)
```

````python
class BESTRecV4(nn.Module):
    def __init__(self, n_users, n_items, gm, ub_init, ib_init):
        super().__init__()
        D = FIELD_DIM

        # Bias
        self.gm = nn.Parameter(torch.tensor(float(gm)), requires_grad=False)
        self.ub = nn.Embedding(n_users, 1)
        self.ub.weight.data = ub_init.unsqueeze(1)
        self.ib = nn.Embedding(n_items, 1)
        self.ib.weight.data = ib_init.unsqueeze(1)

        # CF embeddings (learned from scratch)
        self.u_cf = nn.Embedding(n_users, D)
        self.i_cf = nn.Embedding(n_items, D)
        nn.init.xavier_uniform_(self.u_cf.weight)
        nn.init.xavier_uniform_(self.i_cf.weight)

        # Feature projections -> D-dim
        self.u_text_proj = nn.Sequential(nn.Linear(SBERT_DIM, D), nn.LayerNorm(D))
        self.u_svd_proj = nn.Sequential(nn.Linear(USER_SVD_K, D), nn.LayerNorm(D))
        self.i_text_proj = nn.Sequential(nn.Linear(SBERT_DIM, D), nn.LayerNorm(D))
        self.i_svd_proj = nn.Sequential(nn.Linear(SVD_K, D), nn.LayerNorm(D))
        self.i_num_proj = nn.Sequential(nn.Linear(3, D), nn.LayerNorm(D))

        # DNN
        n_fields = 7
        layers = []
        in_dim = n_fields * D
        for h in DNN_HIDDEN:
            layers.extend([nn.Linear(in_dim, h), nn.ReLU(), nn.BatchNorm1d(h), nn.Dropout(DNN_DROPOUT)])
            in_dim = h
        layers.append(nn.Linear(in_dim, 1))
        self.dnn = nn.Sequential(*layers)

        # DropoutNet state
        self.drop_p = 0.0

    def forward(self, uids, iids, u_text, u_svd, i_title, i_num, i_svd):
        B = uids.size(0)

        # DropoutNet: zero user features during training
        if self.training and self.drop_p > 0:
            mask = torch.rand(B, 1, device=uids.device) < self.drop_p
            u_text = u_text * (~mask).float()
            u_svd = u_svd * (~mask).float()

        # Build field vectors (all D-dim)
        fields = [
            self.u_cf(uids),                              # user CF
            self.u_text_proj(u_text),                     # user text
            self.u_svd_proj(F.normalize(u_svd, dim=-1)),  # user SVD
            self.i_cf(iids),                              # item CF
            self.i_text_proj(i_title),                    # item text
            self.i_svd_proj(F.normalize(i_svd, dim=-1)),  # item SVD
            self.i_num_proj(i_num),                       # item numeric
        ]

        # FM: efficient pairwise interactions
        # 0.5 * (sum(v_i))^2 - sum(v_i^2)   per dimension, then sum
        stacked = torch.stack(fields, dim=1)   # (B, 7, D)
        sum_sq = stacked.sum(dim=1).pow(2)     # (B, D)
        sq_sum = stacked.pow(2).sum(dim=1)     # (B, D)
        fm_out = 0.5 * (sum_sq - sq_sum).sum(dim=-1)  # (B,)

        # DNN
        concat = torch.cat(fields, dim=-1)     # (B, 7*D)
        dnn_out = self.dnn(concat).squeeze(-1)  # (B,)

        # Bias
        bias = self.gm + self.ub(uids).squeeze(-1) + self.ib(iids).squeeze(-1)
        if self.training and self.drop_p > 0:
            # Zero user bias for dropped users
            u_bias = self.ub(uids).squeeze(-1)
            u_bias = u_bias * (~mask.squeeze(-1)).float()
            bias = self.gm + u_bias + self.ib(iids).squeeze(-1)

        pred = torch.clamp(bias + fm_out + dnn_out, 1.0, 5.0)
        return pred

print(f'BESTRecV4 defined. Params: ~{sum(p.numel() for p in BESTRecV4(N_USERS, N_ITEMS, 3.0, torch.zeros(N_USERS), torch.zeros(N_ITEMS)).parameters()):,}')
````

## 9. Training & Evaluation

````python
def get_drop_p(epoch, total):
    ramp_end = int(DROP_RAMP_FRAC * total)
    if epoch < ramp_end:
        return DROP_TARGET * epoch / max(1, ramp_end)
    return DROP_TARGET


def train_epoch(model, opt, sched, scaler, dl):
    model.train()
    total, n = 0.0, 0
    for batch in tqdm(dl, desc='  Train', leave=False):
        uids, iids, u_t, u_s, i_t, i_n, i_s, rat = [x.to(device, non_blocking=True) for x in batch]
        opt.zero_grad(set_to_none=True)
        with autocast(enabled=AMP):
            pred = model(uids, iids, u_t, u_s, i_t, i_n, i_s)
            loss = F.mse_loss(pred, rat)
        scaler.scale(loss).backward()
        scaler.unscale_(opt)
        nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP)
        scaler.step(opt); scaler.update(); sched.step()
        total += loss.item() * rat.size(0); n += rat.size(0)
    return total / n


@torch.no_grad()
def eval_rating(model, dl):
    model.eval()
    ps, ts = [], []
    for batch in tqdm(dl, desc='  Eval', leave=False):
        uids, iids, u_t, u_s, i_t, i_n, i_s, rat = [x.to(device, non_blocking=True) for x in batch]
        with autocast(enabled=AMP):
            pred = model(uids, iids, u_t, u_s, i_t, i_n, i_s)
        ps.append(pred.float().cpu().numpy()); ts.append(rat.cpu().numpy())
    p, t = np.concatenate(ps), np.concatenate(ts)
    return mean_absolute_error(t, p), np.sqrt(mean_squared_error(t, p))


@torch.no_grad()
def eval_ranking(model, test_inters, feats):
    model.eval()
    ut_items = defaultdict(set)
    for i in test_inters: ut_items[i['user_id']].add(i['item_id'])
    ndcg_l, hr_l = [], []
    all_items = set(range(N_ITEMS))
    rng = np.random.RandomState(SEED)
    users = [u for u in ut_items if ut_items[u]][:RANKING_USERS]
    for uid in tqdm(users, desc='  Rank', leave=False):
        for pos in ut_items[uid]:
            neg_pool = list(all_items - ut_items[uid])
            if len(neg_pool) < NEG_SAMPLES: continue
            negs = rng.choice(neg_pool, NEG_SAMPLES, replace=False)
            cands = torch.tensor([pos] + list(negs), dtype=torch.long)
            nc = len(cands)
            uid_t = torch.full((nc,), uid, dtype=torch.long, device=device)
            with autocast(enabled=AMP):
                scores = model(
                    uid_t, cands.to(device),
                    feats['user_text'][uid].unsqueeze(0).expand(nc,-1).to(device),
                    feats['user_svd'][uid].unsqueeze(0).expand(nc,-1).to(device),
                    item_title_emb[cands].to(device),
                    item_num_feat[cands].to(device),
                    feats['item_svd'][cands].to(device)
                )
            scores = scores.float().cpu().numpy()
            ranked = np.argsort(-scores)
            pr = int(np.where(ranked == 0)[0][0])
            hr_l.append(1.0 if pr < TOP_K else 0.0)
            ndcg_l.append(1.0/np.log2(pr+2) if pr < TOP_K else 0.0)
    return {f'NDCG@{TOP_K}': np.mean(ndcg_l) if ndcg_l else 0,
            f'HR@{TOP_K}': np.mean(hr_l) if hr_l else 0}


def run_fold(tag, train_inters, test_inters, do_rank=True):
    print(f"\n{'='*60}")
    print(f'  [{tag}]: {len(train_inters):,} train / {len(test_inters):,} test')
    print(f"{'='*60}")

    feats = compute_fold_features(train_inters, tag)
    train_ds = RecDS(train_inters, feats)
    test_ds = RecDS(test_inters, feats)
    dl_kw = dict(batch_size=BS, num_workers=WORKERS, pin_memory=True,
                 prefetch_factor=4 if WORKERS > 0 else None,
                 persistent_workers=WORKERS > 0)
    train_dl = DataLoader(train_ds, shuffle=True, **dl_kw)
    test_dl = DataLoader(test_ds, shuffle=False, **dl_kw)

    raw = BESTRecV4(N_USERS, N_ITEMS, feats['gm'], feats['ub'], feats['ib']).to(device)
    model = raw
    try:
        if hasattr(torch, 'compile'):
            model = torch.compile(raw, mode='reduce-overhead')
    except: pass

    opt = optim.AdamW(model.parameters(), lr=LR, weight_decay=WD)
    total_steps = EPOCHS * len(train_dl)
    warmup = int(WARMUP_FRAC * total_steps)
    def lr_fn(s):
        if s < warmup: return s / max(1, warmup)
        return 0.5 * (1 + math.cos(math.pi * (s - warmup) / max(1, total_steps - warmup)))
    sched = optim.lr_scheduler.LambdaLR(opt, lr_fn)
    scaler = GradScaler(enabled=AMP)

    best_mae, best_st, pat = float('inf'), None, 0
    for ep in range(EPOCHS):
        t0 = time.time()
        raw.drop_p = get_drop_p(ep, EPOCHS)
        loss = train_epoch(model, opt, sched, scaler, train_dl)
        mae, rmse = eval_rating(model, test_dl)
        mk = ''
        if mae < best_mae:
            best_mae = mae; best_st = copy.deepcopy(raw.state_dict()); pat = 0; mk = ' *'
        else: pat += 1
        print(f'  Ep {ep+1:2d} | loss={loss:.4f} | MAE={mae:.4f} RMSE={rmse:.4f} | dp={raw.drop_p:.2f} | {time.time()-t0:.0f}s{mk}')
        if pat >= PATIENCE: print('  Early stop'); break

    raw.load_state_dict(best_st)
    mae, rmse = eval_rating(model, test_dl)
    res = {'mae': float(mae), 'rmse': float(rmse)}
    print(f'  Best: MAE={mae:.4f} RMSE={rmse:.4f}')

    if do_rank:
        rk = eval_ranking(model, test_inters, feats)
        res.update(rk)
        for k, v in rk.items(): print(f'  {k}: {v:.4f}')

    del model, raw, opt, sched, scaler, train_dl, test_dl
    torch.cuda.empty_cache(); gc.collect()
    return res

print('Training pipeline ready.')
````

## 10. Experiment 1: Warm Evaluation (Per-User Leave-One-Out)

Each test user has training interactions. The model has user features available at test time.
**This is the main result table for the paper.**

````python
print('=' * 60)
print(f'  EXPERIMENT 1: WARM (per-user split) - {DATASET.upper()}')
print('=' * 60)

warm_results = []
for fi, (tr, te) in enumerate(warm_splits):
    r = run_fold(f'warm_f{fi}',
                 [interactions[i] for i in tr],
                 [interactions[i] for i in te],
                 do_rank=True)
    warm_results.append(r)

print(f'\n{"="*60}')
print('WARM RESULTS (main paper table)')
print('='*60)
for m in ['mae', 'rmse', f'NDCG@{TOP_K}', f'HR@{TOP_K}']:
    vals = [r[m] for r in warm_results if m in r]
    if vals:
        mu, sd = np.mean(vals), np.std(vals)
        print(f'  {m:>10s}: {mu:.4f} +/- {sd:.4f}')
````

````python
print('=' * 60)
print(f'  EXPERIMENT 2: COLD (GroupKFold) - {DATASET.upper()}')
print('=' * 60)

cold_results = []
for fi, (tr, te) in enumerate(cold_splits):
    r = run_fold(f'cold_f{fi}',
                 [interactions[i] for i in tr],
                 [interactions[i] for i in te],
                 do_rank=True)
    cold_results.append(r)

print(f'\n{"="*60}')
print('COLD RESULTS')
print('='*60)
for m in ['mae', 'rmse', f'NDCG@{TOP_K}', f'HR@{TOP_K}']:
    vals = [r[m] for r in cold_results if m in r]
    if vals:
        print(f'  {m:>10s}: {np.mean(vals):.4f} +/- {np.std(vals):.4f}')
````

## 12. Experiment 3: Baselines

````python
def svd_baseline(train_inters, test_inters, k=128):
    gs, gn = 0.0, 0
    us, uc = defaultdict(float), defaultdict(int)
    ist, ic = defaultdict(float), defaultdict(int)
    for i in train_inters:
        r = i['rating']; gs += r; gn += 1
        us[i['user_id']] += r; uc[i['user_id']] += 1
        ist[i['item_id']] += r; ic[i['item_id']] += 1
    gm = gs / gn
    rows, cols, vals = [], [], []
    for i in train_inters:
        ub = (us[i['user_id']]/uc[i['user_id']])-gm if uc[i['user_id']]>0 else 0
        ib = (ist[i['item_id']]/ic[i['item_id']])-gm if ic[i['item_id']]>0 else 0
        rows.append(i['user_id']); cols.append(i['item_id']); vals.append(i['rating']-gm-ub-ib)
    mat = csr_matrix((vals,(rows,cols)), shape=(N_USERS, N_ITEMS))
    ak = min(k, min(N_USERS,N_ITEMS)-1, len(set(rows))-1)
    if ak < 1: U, V = np.zeros((N_USERS,1)), np.zeros((1,N_ITEMS))
    else:
        svd = TruncatedSVD(n_components=ak, random_state=SEED)
        U = svd.fit_transform(mat); V = svd.components_
    preds = []
    for i in test_inters:
        ub = (us[i['user_id']]/uc[i['user_id']])-gm if uc[i['user_id']]>0 else 0
        ib = (ist[i['item_id']]/ic[i['item_id']])-gm if ic[i['item_id']]>0 else 0
        s = U[i['user_id']]@V[:,i['item_id']] if uc[i['user_id']]>0 and ak>=1 else 0
        preds.append(np.clip(gm+ub+ib+s, 1, 5))
    tgts = [i['rating'] for i in test_inters]
    return {'mae': mean_absolute_error(tgts,preds), 'rmse': np.sqrt(mean_squared_error(tgts,preds))}


def neumf_baseline(train_inters, test_inters, epochs=30):
    class NeuMF(nn.Module):
        def __init__(self):
            super().__init__()
            self.ug = nn.Embedding(N_USERS, 64); self.ig = nn.Embedding(N_ITEMS, 64)
            self.um = nn.Embedding(N_USERS, 64); self.im = nn.Embedding(N_ITEMS, 64)
            self.mlp = nn.Sequential(nn.Linear(128,128),nn.ReLU(),nn.Dropout(0.1),nn.Linear(128,64),nn.ReLU())
            self.out = nn.Linear(128, 1)
            for e in [self.ug,self.ig,self.um,self.im]: nn.init.xavier_uniform_(e.weight)
        def forward(self, u, i):
            g = self.ug(u)*self.ig(i); m = self.mlp(torch.cat([self.um(u),self.im(i)],-1))
            return torch.clamp(self.out(torch.cat([g,m],-1)).squeeze(-1)+3.0, 1, 5)

    model = NeuMF().to(device); opt = optim.Adam(model.parameters(), lr=1e-3)
    tu = torch.tensor([i['user_id'] for i in train_inters], dtype=torch.long)
    ti = torch.tensor([i['item_id'] for i in train_inters], dtype=torch.long)
    tr = torch.tensor([i['rating'] for i in train_inters], dtype=torch.float32)
    eu = torch.tensor([i['user_id'] for i in test_inters], dtype=torch.long)
    ei = torch.tensor([i['item_id'] for i in test_inters], dtype=torch.long)
    er = torch.tensor([i['rating'] for i in test_inters], dtype=torch.float32)
    bm, bs, bp = float('inf'), None, 0
    for ep in range(epochs):
        model.train(); perm = torch.randperm(len(tr))
        for s in range(0, len(tr), BS):
            idx = perm[s:s+BS]
            loss = F.mse_loss(model(tu[idx].to(device),ti[idx].to(device)), tr[idx].to(device))
            opt.zero_grad(set_to_none=True); loss.backward(); opt.step()
        model.eval()
        with torch.no_grad():
            ps = [model(eu[s:s+BS].to(device),ei[s:s+BS].to(device)).cpu().numpy() for s in range(0,len(er),BS)]
        mae = mean_absolute_error(er.numpy(), np.concatenate(ps))
        if mae < bm: bm = mae; bs = copy.deepcopy(model.state_dict()); bp = 0
        else: bp += 1
        if bp >= 8: break
    model.load_state_dict(bs); model.eval()
    with torch.no_grad():
        ps = [model(eu[s:s+BS].to(device),ei[s:s+BS].to(device)).cpu().numpy() for s in range(0,len(er),BS)]
    pa = np.concatenate(ps)
    del model; torch.cuda.empty_cache()
    return {'mae': mean_absolute_error(er.numpy(),pa), 'rmse': np.sqrt(mean_squared_error(er.numpy(),pa))}


# Run baselines on warm fold 0
print('=' * 60)
print(f'  BASELINES (warm fold 0) - {DATASET.upper()}')
print('=' * 60)
tr0 = [interactions[i] for i in warm_splits[0][0]]
te0 = [interactions[i] for i in warm_splits[0][1]]

# Global mean
gm0 = np.mean([i['rating'] for i in tr0])
gm_mae = mean_absolute_error([i['rating'] for i in te0], np.full(len(te0), gm0))
print(f'  Global Mean: MAE={gm_mae:.4f}')

# Item mean from training
im0 = defaultdict(list)
for i in tr0: im0[i['item_id']].append(i['rating'])
imd0 = {k: np.mean(v) for k, v in im0.items()}
im_preds = [imd0.get(i['item_id'], gm0) for i in te0]
im_mae = mean_absolute_error([i['rating'] for i in te0], im_preds)
print(f'  Item Mean:   MAE={im_mae:.4f}')

svd_r = svd_baseline(tr0, te0)
print(f'  SVD:         MAE={svd_r["mae"]:.4f} RMSE={svd_r["rmse"]:.4f}')

nmf_r = neumf_baseline(tr0, te0)
print(f'  NeuMF:       MAE={nmf_r["mae"]:.4f} RMSE={nmf_r["rmse"]:.4f}')

v4_f0 = warm_results[0]
print(f'  BEST-Rec v4: MAE={v4_f0["mae"]:.4f} RMSE={v4_f0["rmse"]:.4f}')
if svd_r['mae'] > 0:
    print(f'  Improvement over SVD: {(svd_r["mae"]-v4_f0["mae"])/svd_r["mae"]*100:.1f}%')
````

## 13. Ablation Study

````python
print('=' * 60)
print(f'  ABLATION (warm fold 0) - {DATASET.upper()}')
print('=' * 60)

abl_tr = [interactions[i] for i in warm_splits[0][0]]
abl_te = [interactions[i] for i in warm_splits[0][1]]
ablation = {'Full v4': warm_results[0]}
print(f'  Full v4: MAE={warm_results[0]["mae"]:.4f}')

# No DropoutNet
old_drop = DROP_TARGET
DROP_TARGET = 0.0
ablation['No Dropout'] = run_fold('abl_nodrop', abl_tr, abl_te, do_rank=False)
DROP_TARGET = old_drop

# No text features (zero out text inputs)
# Simplest: set SBERT_DIM projections to zero by modifying features
print('\n  --- No Text Features ---')
feats_notext = compute_fold_features(abl_tr, 'abl_notext')
feats_notext['user_text'] = torch.zeros_like(feats_notext['user_text'])
# Run with zeroed text
train_ds_nt = RecDS(abl_tr, feats_notext)
test_ds_nt = RecDS(abl_te, feats_notext)
dl_kw = dict(batch_size=BS, num_workers=WORKERS, pin_memory=True,
             prefetch_factor=4 if WORKERS>0 else None, persistent_workers=WORKERS>0)
train_dl_nt = DataLoader(train_ds_nt, shuffle=True, **dl_kw)
test_dl_nt = DataLoader(test_ds_nt, shuffle=False, **dl_kw)

raw_nt = BESTRecV4(N_USERS, N_ITEMS, feats_notext['gm'], feats_notext['ub'], feats_notext['ib']).to(device)
# Also zero out item title embeddings in the dataset
orig_title = item_title_emb.clone()
item_title_emb.zero_()

opt_nt = optim.AdamW(raw_nt.parameters(), lr=LR, weight_decay=WD)
ts_nt = EPOCHS * len(train_dl_nt)
wu_nt = int(WARMUP_FRAC * ts_nt)
sched_nt = optim.lr_scheduler.LambdaLR(opt_nt, lambda s: s/max(1,wu_nt) if s<wu_nt else 0.5*(1+math.cos(math.pi*(s-wu_nt)/max(1,ts_nt-wu_nt))))
scaler_nt = GradScaler(enabled=AMP)
bm_nt, bs_nt, bp_nt = float('inf'), None, 0
for ep in range(EPOCHS):
    raw_nt.drop_p = get_drop_p(ep, EPOCHS)
    train_epoch(raw_nt, opt_nt, sched_nt, scaler_nt, train_dl_nt)
    mae_nt, rmse_nt = eval_rating(raw_nt, test_dl_nt)
    if mae_nt < bm_nt: bm_nt = mae_nt; bs_nt = copy.deepcopy(raw_nt.state_dict()); bp_nt = 0
    else: bp_nt += 1
    if bp_nt >= PATIENCE: break
raw_nt.load_state_dict(bs_nt)
mae_nt, rmse_nt = eval_rating(raw_nt, test_dl_nt)
ablation['No Text'] = {'mae': float(mae_nt), 'rmse': float(rmse_nt)}
print(f'  No Text: MAE={mae_nt:.4f} RMSE={rmse_nt:.4f}')

# Restore
item_title_emb.copy_(orig_title)
del raw_nt, opt_nt, sched_nt, scaler_nt, train_dl_nt, test_dl_nt
torch.cuda.empty_cache(); gc.collect()

print(f'\n{"="*60}')
print('ABLATION SUMMARY')
print('='*60)
for name, res in ablation.items():
    print(f'  {name:>15s}: MAE={res["mae"]:.4f} RMSE={res["rmse"]:.4f}')
````

## 14. Results Summary & Export

````python
def jf(o):
    if isinstance(o, (np.floating,)): return float(o)
    if isinstance(o, (np.integer,)): return int(o)
    if isinstance(o, np.ndarray): return o.tolist()
    if isinstance(o, dict): return {k: jf(v) for k, v in o.items()}
    if isinstance(o, list): return [jf(v) for v in o]
    return o

print('\n' + '#' * 65)
print(f'  BEST-Rec v4.0 (DeepFM) RESULTS: {DATASET.upper()}')
print('#' * 65)

print('\n=== WARM (per-user split, main table) ===')
for m in ['mae', 'rmse', f'NDCG@{TOP_K}', f'HR@{TOP_K}']:
    vals = [r[m] for r in warm_results if m in r]
    if vals:
        mu, sd = np.mean(vals), np.std(vals)
        ci = 1.96 * sd / np.sqrt(len(vals))
        print(f'  {m:>10s}: {mu:.4f} +/- {sd:.4f} (95% CI: [{mu-ci:.4f}, {mu+ci:.4f}])')

print('\n=== COLD (GroupKFold) ===')
for m in ['mae', 'rmse', f'NDCG@{TOP_K}', f'HR@{TOP_K}']:
    vals = [r[m] for r in cold_results if m in r]
    if vals:
        print(f'  {m:>10s}: {np.mean(vals):.4f} +/- {np.std(vals):.4f}')

print('\n=== BASELINES (warm fold 0) ===')
print(f'  Global Mean: MAE={gm_mae:.4f}')
print(f'  Item Mean:   MAE={im_mae:.4f}')
print(f'  SVD:         MAE={svd_r["mae"]:.4f}')
print(f'  NeuMF:       MAE={nmf_r["mae"]:.4f}')

print('\n=== ABLATION (warm fold 0) ===')
for name, res in ablation.items():
    print(f'  {name:>15s}: MAE={res["mae"]:.4f}')

# Warm vs cold degradation
w_mae = np.mean([r['mae'] for r in warm_results])
c_mae = np.mean([r['mae'] for r in cold_results])
print(f'\nWarm->Cold degradation: {(c_mae-w_mae)/w_mae*100:.1f}%')

# Save
all_res = jf({
    'dataset': DATASET, 'version': 'v4.0',
    'warm': warm_results, 'cold': cold_results,
    'baselines': {'global_mean': {'mae': float(gm_mae)}, 'item_mean': {'mae': float(im_mae)},
                  'svd': svd_r, 'neumf': nmf_r},
    'ablation': {k: v for k, v in ablation.items()},
})
out = os.path.join(V4_CACHE, 'v4_results.json')
json.dump(all_res, open(out, 'w'), indent=2)
print(f'\nSaved to {out}')
print('Done! Run on 3 more datasets, then share all 4 JSONs.')
````
